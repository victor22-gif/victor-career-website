```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8) #adjusts the configuration of the plots we will create

# Read in the data
df = pd.read_csv('/content/drive/MyDrive/movies.csv')

# Set display option
pd.set_option("display.max_rows", None)
```

```python
# Lets look at the data
df.head()
```

```python
# checking for missing data
for col in df.columns:
  pct_missing = np.mean(df[col].isnull())
  print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
# Drop missing values
df = df.dropna()
```

```python
# checking again for missing data
for col in df.columns:
  pct_missing = np.mean(df[col].isnull())
  print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
# check for daa types of features
df.dtypes
```

```python
# Change data type of 'budget' and 'gross' to numeric values.

df['budget'] = df['budget'].astype('int64')
df['gross'] = df['gross'].astype('int64')
```

```python
# Creating year column
import re

df['yearcorrect'] = df['released'].str.extract(r'(\d{4})')
df['yearcorrect'] = df['yearcorrect'].astype('int64')

df
```

```python

df.sort_values(by=['gross'], inplace=False, ascending=False)
```

```python
# Drop any duplicates
df.drop_duplicates()
```

```python
# Checking for correlation with 'budget' vs 'gross'

plt.scatter(x=df['budget'], y=df['gross'])
plt.title('Budget vs Gross Earnings')
plt.xlabel('Gross Earnings')
plt.ylabel('Budget for Film')
plt.show()
```

```python
#

sns.regplot(x='budget', y='gross', data=df, scatter_kws={"color":"red"}, line_kws={"color":"blue"})
```

```python
# Check correlation

df.select_dtypes(include=[np.number]).corr()
```

```python
# Visualize the correlation matrix
corr_matrix = df.select_dtypes(include=[np.number]).corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title('Correlation Matrix')
plt.show();
```

```python
## High correlation between gross and budget
```

```python
# Company
df["company"].head()
```

```python
# Make objects categorical
numerised_df = df.copy()
for col_name in numerised_df.columns:
    if(numerised_df[col_name].dtype == 'object'):
        numerised_df[col_name] = numerised_df[col_name].astype('category')
        numerised_df[col_name] = numerised_df[col_name].cat.codes

numerised_df
```

```python
# checking again for missing data
for col in numerised_df.columns:
  pct_missing = np.mean(numerised_df[col].isnull())
  print('{} - {}%'.format(col, round(pct_missing*100)))
```

```python
# Visualize the correlation matrix
corr_matrix = numerised_df.corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title('Correlation Matrix')
plt.show();
```

```python
corr_pair = corr_matrix.unstack()

#Sort values
corr_sorted = corr_pair.sort_values()
```

```python
# View sorted values
corr_sorted
```

```python
high_correlation = corr_sorted[(corr_sorted)>0.5]

high_correlation
```
